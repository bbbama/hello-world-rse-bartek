## Quick example
Let's define the function directly here for a quick test.

```python
from pint import UnitRegistry

ureg = UnitRegistry()

def calculate_speed(dist, dist_unit, time, time_unit):
    d = dist * ureg(dist_unit)
    t = time * ureg(time_unit)
    speed = d / t
    return speed.to(ureg.meter / ureg.second)

# Test: 100 meters in 10 seconds
print("Speed:", calculate_speed(100, 'm', 10, 's'))
```

## Different units
The cool thing is that it handles unit conversion (like km/h to m/s) for us.

```python
v = calculate_speed(36, "km", 1, "hour")
print(f"36 km/h is {v}")
```
